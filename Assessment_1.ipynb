```python
#31 Armstrong Numbers
armstrong = []

for number in range(1, 10000):
    temp = number
    digits = 0

    while temp > 0:
        digits += 1
        temp //= 10

    temp = number
    total = 0

    while temp > 0:
        digit = temp % 10
        total += digit ** digits
        temp //= 10

    if total == number:
        armstrong.append(number)

print("Armstrong numbers:")

for i in range(len(armstrong)):
    if i > 0:
        print(", ", end="")
    print(armstrong[i], end="")

print()
print("Count:", len(armstrong))
```

```python
#33 Prime check + Sum of Primes
N = int(input("Enter N: "))
if N < 2:
    print("Not prime")
else:
    is_prime = True
    i = 2

    while i * i <= N:
        if N % i == 0:
            is_prime = False
            break
        i += 1

    if is_prime:
        print(N, "is prime")
    else:
        print(N, "is not prime")

prime_sum = 0

for number in range(2, N + 1):
    prime = True
    i = 2

    while i * i <= number:
        if number % i == 0:
            prime = False
            break
        i += 1

    if prime:
        prime_sum += number

print("Sum of primes:", prime_sum)
```

```python
#Q34. Reverse Integer + Palindrome
number = int(input("Enter a number: "))

if number < 0:
    print("Reversed:", number)
    print("Not a palindrome")
else:
    original = number
    reverse = 0

    while number > 0:
        digit = number % 10
        reverse = reverse * 10 + digit
        number //= 10

    print("Reversed:", reverse)

    if original == reverse:
        print("Palindrome")
    else:
        print("Not a palindrome")
```

```python
#Q36. String Scanner — Single Pass
sentence = input("Enter a sentence: ")

total_characters = 0
words = 0
vowels = 0
consonants = 0
digits = 0
spaces = 0
special = 0

inside_word = False

for ch in sentence:
    total_characters += 1

    if ch.isspace():
        spaces += 1
        inside_word = False

    elif ch.isalpha():
        if ch.lower() in "aeiou":
            vowels += 1
        else:
            consonants += 1

        if not inside_word:
            words += 1
            inside_word = True

    elif ch.isdigit():
        digits += 1

        if not inside_word:
            words += 1
            inside_word = True

    else:
        special += 1

print("Total characters:", total_characters)
print("Total words:", words)
print("Vowels:", vowels)
print("Consonants:", consonants)
print("Digits:", digits)
print("Spaces:", spaces)
print("Special characters:", special)
```

```python
#Q37. Student Marks
marks = list(map(int, input("Enter 10 marks: ").split()))

maximum = marks[0]
minimum = marks[0]
total = 0

for mark in marks:
    if mark > maximum:
        maximum = mark

    if mark < minimum:
        minimum = mark

    total += mark

average = total / len(marks)

above_average = 0

for mark in marks:
    if mark > average:
        above_average += 1

highest = maximum
second_highest = None

for mark in marks:
    if mark < highest:
        if second_highest is None or mark > second_highest:
            second_highest = mark

print("Maximum:", maximum)
print("Minimum:", minimum)
print("Average:", average)
print("Above average:", above_average)
print("Second highest:", second_highest)
```

```python
#Q38. Word Frequency Analyzer
review = input("Enter a review: ")

words = review.lower().split()
frequency = {}

for word in words:
    word = word.strip(".,!?;:'\"()[]{}")

    if word != "":
        if word not in frequency:
            frequency[word] = 0

        frequency[word] += 1

sorted_words = sorted(frequency, key=frequency.get, reverse=True)

print("Word frequencies:")

for word in sorted_words:
    print(word, ":", frequency[word])

if len(sorted_words) > 0:
    print("Most frequent word:", sorted_words[0])
    print("Count:", frequency[sorted_words[0]])
```

```python
#Q39. Indian States CRUD
states = {
    "Karnataka": "Bengaluru",
    "Maharashtra": "Mumbai",
    "Tamil Nadu": "Chennai",
    "Kerala": "Thiruvananthapuram",
    "Goa": "Panaji"
}

while True:
    print("\n1. Add")
    print("2. Update")
    print("3. Delete")
    print("4. Search")
    print("5. Display")
    print("6. Exit")

    choice = input("Enter choice: ")

    if choice == "1":
        state = input("Enter state: ")
        capital = input("Enter capital: ")
        states[state] = capital
        print("State added.")

    elif choice == "2":
        state = input("Enter state: ")

        if state in states:
            capital = input("Enter new capital: ")
            states[state] = capital
            print("Capital updated.")
        else:
            print("State not found.")

    elif choice == "3":
        state = input("Enter state: ")

        if state in states:
            del states[state]
            print("State deleted.")
        else:
            print("State not found.")

    elif choice == "4":
        state = input("Enter state: ")

        if state in states:
            print("Capital:", states[state])
        else:
            print("Not found")

    elif choice == "5":
        for state in sorted(states):
            print(state, ":", states[state])

    elif choice == "6":
        print("Program ended.")
        break

    else:
        print("Invalid choice.")
```

```python
#Q42. Duplicate Numbers
numbers = [10, 20, 30, 10, 40, 50, 20, 60, 70, 30,
           80, 90, 100, 40, 110, 120, 130, 50, 140, 150]

unique_numbers = set(numbers)

duplicates = []
exactly_once = []

for number in unique_numbers:
    count = numbers.count(number)

    if count > 1:
        duplicates.append(number)
    elif count == 1:
        exactly_once.append(number)

print("Unique numbers:", unique_numbers)
print("Numbers appearing more than once:", duplicates)
print("Numbers appearing exactly once:", exactly_once)
```

```python
#Q45. Caesar Cipher
message = input("Enter message: ")
shift = int(input("Enter shift: "))

result = ""

for ch in message:
    if "A" <= ch <= "Z":
        new_ch = chr((ord(ch) - ord("A") + shift) % 26 + ord("A"))
        result += new_ch

    elif "a" <= ch <= "z":
        new_ch = chr((ord(ch) - ord("a") + shift) % 26 + ord("a"))
        result += new_ch

    else:
        result += ch

print("Encrypted message:", result)
```

```python
#Q47. Recursive Power
def power(base, exp):
    if exp == 0:
        return 1

    if exp < 0:
        return 1 / power(base, -exp)

    return base * power(base, exp - 1)


print(power(2, 10))
print(power(5, 0))
print(power(3, -2))
```
