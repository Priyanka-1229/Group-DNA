```python
#Whatsapp chat parser
chat = []
for line in lines:
    if " - " in line and ": " in line:
        parts = line.split(" - ", 1)
        date_time = parts[0]
        sender_text = parts[1]
        sender, text = sender_text.split(": ", 1)
        chat.append([date_time, sender, text.strip()])
print("Total messages:", len(chat))
print("First message:", chat[0])
```

```python
#Group Overview
people = []
message_counts = {}
for message in chat:
    sender = message[1]
    if sender not in people:
        people.append(sender)
        message_counts[sender] = 0
    message_counts[sender] += 1
print("Total messages:", len(chat))
print("Total participants:", len(people))
for person in people:
    print(person, ":", message_counts[person])
```

```python
#Most Active Day and Hour
day_count = {}
hour_count = {}
for message in chat:
    date_time = message[0]
    date = date_time.split(",")[0]
    time = date_time.split(",")[1].strip()
    hour = int(time.split(":")[0])
    if date not in day_count:
        day_count[date] = 0
    day_count[date] += 1
    if hour not in hour_count:
        hour_count[hour] = 0
    hour_count[hour] += 1
active_day = max(day_count, key=day_count.get)
active_hour = max(hour_count, key=hour_count.get)
print("Most active day:", active_day)
print("Messages on that day:", day_count[active_day])
print("Most active hour:", active_hour)
print("Messages in that hour:", hour_count[active_hour])
```

```python
#NumPy Activity Heatmap
from datetime import datetime
import numpy as np
activity = np.zeros((7, 24), dtype=int)
for message in chat:
    date_time = message[0]
    date = date_time.split(",")[0]
    time = date_time.split(",")[1].strip()
    day = datetime.strptime(date, "%d/%m/%y").weekday()
    hour = int(time.split(":")[0])
    activity[day][hour] += 1
print("Activity Heatmap")
print(activity)
```

```python
#Top Words
word_count = {}
for message in chat:
    words = message[2].lower().split()
    for word in words:
        word = word.strip(".,!?\"'()[]{}:")
        if len(word) > 2:
            if word not in word_count:
                word_count[word] = 0
            word_count[word] += 1
top_words = sorted(word_count, key=word_count.get, reverse=True)
print("Top 20 words:")
for word in top_words[:20]:
    print(word, ":", word_count[word])
```

```python
#Response Speed & Silent Streaks
from datetime import datetime
response_times = []
silent_streaks = []
previous_time = None
for message in chat:
    current_time = datetime.strptime(message[0], "%d/%m/%y, %H:%M")

    if previous_time is not None:
        difference = (current_time - previous_time).total_seconds() / 60

        if difference <= 60:
            response_times.append(difference)
        else:
            silent_streaks.append(difference)
    previous_time = current_time
if len(response_times) > 0:
    average_response = sum(response_times) / len(response_times)
else:
    average_response = 0
if len(silent_streaks) > 0:
    longest_silence = max(silent_streaks)
else:
    longest_silence = 0
print("Average response time:", round(average_response, 2), "minutes")
print("Longest silent streak:", round(longest_silence / 60, 2), "hours")
```
